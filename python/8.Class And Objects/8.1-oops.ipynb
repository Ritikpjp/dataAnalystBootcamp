```python
## dinfining a clss with instance methods and instance variables
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def bark(self): ## this is an instance method and it is used to define the behaviour of the class dog
                    ##the self keyword is used here so that we can access the instance variables of the class inside this instance method
        print(f"{self.name} says woof woof") ## this is the behaviour of the class dog and it

dog1=dog("buddy",5)
dog1.bark() ## here we are calling the instance method bark of the object dog1
dog2=dog("max",3)
dog2.bark() ## here we are calling the instance method bark of the object dog2
```

```python
## modeling a bnak account
class bankAccount:
    def __init__(self,accountNumber,balance):
        self.accountNumber=accountNumber
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print(f"Deposited {amount} to account {self.accountNumber}. New balance is {self.balance}")
    def withdraw(self,amount):
        if amount>self.balance:
            print(f"Insufficient balance in account {self.accountNumber}. Current balance is {self.balance}")
        else:
            self.balance-=amount
            print(f"Withdrew {amount} from account {self.accountNumber}. New balance is {self.balance}")
    def getBalance(self):
        return self.balance
owner1=bankAccount("123456789",1000)
print (owner1.balance)
owner1.withdraw(200)
owner1.deposit(500)
owner1.withdraw(200)
print(f"Final balance for account {owner1.accountNumber}: {owner1.getBalance()}")
print (owner1.balance)
```
