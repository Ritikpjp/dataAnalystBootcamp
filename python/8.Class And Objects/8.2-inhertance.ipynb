##### Inheritance in python
###### inhertance is a fundamental concept in oops that allows the class to inherit he attributes and methods from  another class.this lessen covers inhertance in single inheritance ,multiple inheritance demoontrating how to create them and use them in pyhton

```python
## Inheritance (Single Inheritance)
## Parent class
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car ")
```

```python
car1=Car(4,5,"petrol")
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype)
        self.is_selfdriving=is_selfdriving

    def driving(self):
        print(f"Tesla supports self driving : {self.is_selfdriving}")
```

```python
tesla1=Tesla(4,5,"electric",True)
tesla1.driving()
print(tesla1.doors)
car1=Car(4,5,"electric")
print(car1.doors)##her we can not use self.doors because 
```

```python
tesla1.drive()
```

```python
### Multiple Inheritance
## When a class inherits from more than one base class.
## Base class 1
class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print("Subclass must implement this method")

## BAse class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner


##Derived class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)

    def speak(self):
        return f"{self.name} say woof"
    

## Create an object
dog=Dog("Buddy","Krish")
print(dog.speak())
print(f"Owner:{dog.owner}")
```
