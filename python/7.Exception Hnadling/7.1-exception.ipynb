###### ZeroDivisionError:Dividing by zero
###### FileNotFoundError:file not found
###### Value Error:invalid value

```python
try:
    num=int(input("enter a number"))
    result=10/num
except ValueError:
    print("not a valid number")
except ZeroDivisionError:
    print("enter demnomintor greater than 0")
except Exception as ex:
    print(ex)
```

```python
## try ,except ,else block
try:
    num=int(input("enter a number"))
    result=10/num
except ValueError:
    print("not a valid number")
except ZeroDivisionError:
    print("enter demnomintor greater than 0")
except Exception as ex:
    print(ex)
else:##after try gets executed successfully it had to accept this
    print(f"the result is {result}")
    ## whenever some exception occurs it gets handled in the except block
    ## but if there is no exception it gets executed and shows the result
```

```python
## try ,except,else and finally
try:
    num=int(input("enter a number"))
    result=10/num
except ValueError:
    print("not a valid number")
except ZeroDivisionError:
    print("enter demnomintor greater than 0")
except Exception as ex:
    print(ex)
else:## this wil get executed only when there is no error in the try
    print(f"the result is {result}")
finally:
    print("execution completed")

```
