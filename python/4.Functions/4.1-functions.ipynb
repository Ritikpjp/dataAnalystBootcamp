```python
## variable length arguments
## positional and keyWORD ARGUMENTS
def print_numbers(*args):
    for number in args:
        print(number)
        
    
```

```python
print_numbers(1,2,3,4,5,6,7,7,8,"jdkjs")
```

```python
## keyword arguments## all the keys or parameters will be in the form of key value pair
def print_details(**kwargs):
    for key ,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="keisr",age="32",country="india")
```

```python
## IN THIS WE ARE TRYING TP
def print_details(*args,**kwargs):
    for number in args:
        print(f"positional arg:{number}")
    for key ,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,4,5,"ehy",name="keisr",age="32",country="india")## positional agruments can not appear after keyword arguments
```
