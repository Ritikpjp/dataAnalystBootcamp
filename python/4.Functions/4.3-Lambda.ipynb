###Lambda Function in python

###### lambda function are smalll anonymos function defined using the lambda keyword.They can have any nimber of arguments but only one expression.They are commonly used for short operations or as arguments to higher-order functions,a function without name

```python
#Syntax
lambda arguments : expression
```

```python
def addition (x, y):
    return x + y

# Lambda function for addition
addition_lambda = lambda x, y: x + y
```

```python
addition(5, 4)  # Output: 9
addition_lambda(5, 3)  # Output: 8
```

```python
addition=lambda a,b:a+b ##lambda is an annonymous function, it does not have a name, we can assign it to a variable like addition
type(addition) ##it is a function
print(addition(5, 4)) ##Output: 9
```

```python
def even(number):
    return number % 2 == 0
even(24)
```

```python
even1=lambda number:number%2==0
even1(24)
```

```python
def add(x, y, z):
    return x + y + z
add(1, 2, 3) ##Output: 6
```

```python
addition2=lambda x,y,z:x+y+z
addition2(1, 2, 3) ##Output: 6
```

```python
## map()
##class map(func: (_T1@__new__) -> _S@map, iterable: Iterable[_T1@__new__],  /)
## map uses a function to an iterable and returns a map object (which is an iterator) that yields the results of applying the function to each item of the iterable.
## uses lazy loading technique, it does not compute the results immediately, it computes the results on demand when you iterate over the map object.
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(squared) ##Output: [1, 4, 9, 16, 25]  
```
