```python
### read a file line by line
with open('example.txt', 'r') as f:
    for line in f:# iterate over each line in the file
        #print(line)# Note: this will include the newline character at the end of each line
        print(line.strip())# print the line without leading/trailing whitespace
```

```python
## writing a file (overwriting)
with open('example.txt', 'w') as f:# open the file for writing (this will overwrite the file if it already exists)
    f.write('This is a new file.\n')# write a line to the file
    f.write('This file was created using Python.\n')# write another line to the file
```

```python
##write a file without overwriting (append mode)
with open('example.txt', 'a') as f:# open the file for appending (this will add to the end of the file if it already exists)
    f.write('This line was added to the file.\n')# write a line to the file  ,\n is used to add a new line at the end of the line being written
       
```

```python
##writing list of lines to a file
lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']# create a list of lines to write to the file
with open('example.txt', 'a') as f:# open the file for writing (this will overwrite the file if it already exists)
    f.writelines(lines)# write the list of lines to the file
```

```python
## binary files- binary files are files that contain data in a format that is not human-readable, such as images, audio files, or executable files. When working with binary files in Python, you need to open the file in binary mode by using the 'b' flag in the open() function.
## in binary files we usually udates bytes instead of strings, for example when writing to a binary file we need to encode the string to bytes using the encode() method, and when reading from a binary file we need to decode the bytes back to a string using the decode() method.
## writing to a binary file
data =b'\x00\x01\x02\x03\x04\x05'# create some binary data to write to the file
with open('example.bin', 'wb') as f:# open the file for writing in binary mode
    f.write(data)# write the binary data to the file
```

```python
## reading from a binary file
with open('example.bin', 'rb') as f:# open the file for reading in binary mode
    data = f.read()# read the binary data from the file
    print(data)# print the binary data (this will be in bytes format)
```

```python
## i want to read the content from a source text file and write to a destination text file
## copying a text file
with open('example.txt', 'r') as src, open('destination.txt', 'w') as dst:# open the source file for reading and the destination file for writing
    content = src.read()# read the content from the source file
with open('destination.txt', 'w') as dst:# open the destination file for writing
    dst.write(content)# write the content to the destination file
## wha the above code does is it first reads the content from the source file and stores it in a variable, then it opens the destination file for writing and writes the content to it. This will create a copy of the source file with the name 'destination.txt'.
## it will automatically create the destination file if it does not exist, and if it already exists it will overwrite the content of the destination file with the content of the source file.
```

```python
##read a text file and count the number of lines,words, and characters in the file
def count_file_content(filename):
    with open(filename, 'r') as f:# open the file for reading
        lines = f.readlines()# read all lines from the file into a list
        num_lines = len(lines)# count the number of lines
        num_words = sum(len(line.split()) for line in lines)# count the number of words by splitting each line into words and summing their lengths
        num_chars = sum(len(line) for line in lines)# count the number of characters by summing the length of each line
    return num_lines, num_words, num_chars
filename = 'example.txt'
lines, words, chars = count_file_content(filename)
print(f'Number of lines: {lines}')
print(f'Number of words: {words}')
print(f'Number of characters: {chars}')
```

```python
## writing and then reading a file using the with statement
with open('example.txt', 'w+') as f:# open the file for reading and writing (this will overwrite the file if it already exists)
    #w+=is used to open the file for both reading and writing, and it will overwrite the file if it already exists. If the file does not exist, it will create a new file.
    f.write('This is a new file.\n')# write a line to
    f.write('This file was created using Python.\n')# write another line to the file
    ## move the file pointer back to the beginning of the file so we can read the content we just wrote
    f.seek(0)# move the file pointer back to the beginning of the file
    # read the content from the file    
    content = f.read()# read the content from the file
    print(content)# print the content
```
