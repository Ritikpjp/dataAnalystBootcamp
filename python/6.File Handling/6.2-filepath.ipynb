```python
## create a new directory
import os
new_directory = 'package'# specify the name of the new directory
os.mkdir(new_directory)# create the new directory 
print(f'Directory "{new_directory}" created successfully.')
```

```python
## listing files and directories
items = os.listdir('.')# list all items in the current directory
#. is used to specify the current directory
print(items)
## another way is
for item in os.listdir('.'):
    print(item)
     ## another is
for item in items:
    print(item)
```

```python
### joining paths
dir_name = 'folder'# specify the directory name
file_name = 'file.txt'# specify the file name
## so now i want to create a path to the file inside the folder, i can use os.path.join() to join the directory name and the file name
file_path = os.path.join(dir_name, file_name)# join the directory name and the file name to create a file path,these two arguments will be joined together to create a file path that points to the file.txt inside the folder directory. The os.path.join() function takes care of adding the appropriate path separator (such as '/' on Unix-based systems or '\' on Windows) between the directory name and the file name, ensuring that the resulting file path is valid for the operating system being used.they are relative paths because they are relative to the current working directory. If you want to create an absolute path, you can use os.path.abspath() to get the absolute path of the file.
print(file_path)# print the file path   
```

```python
### joining paths
dir_name = 'folder'# specify the directory name
file_name = 'file.txt'# specify the file name
## so now i want to create a path to the file inside the folder, i can use os.path.join() to join the directory name and the file name
file_path = os.path.join(os.getcwd(), dir_name, file_name)# join the directory name and the file name to create a file path
print(file_path)# print the file path
```

```python
path='example1.txt'# specify the file path
if os.path.exists(path):# check if the file exists
    print(f'The file "{path}" exists.')# print a message if the file exists
else:
    print(f'The file "{path}" does not exist.')# print a message if the file does not exist
```

```python
## another way to check if a file exists is to use os.path.isfile() function, which checks if the specified path is a file and exists. Here is an example:
## checkking if path is a file or directory
import os
path = 'example.txt'  # specify the file path
if os.path.isfile(path):
    print(f'The path "{path}" is a file.')
elif os.path.isdir(path):
    print(f'The path "{path}" is a directory.')
else:
    print(f'The path "{path}" does not exist or is not a file or directory.') 
     
      
```

```python
## getting the absolute path
relative_path = 'example.txt'  
# Stores the relative path (filename in the current working directory)

absolute_path = os.path.abspath(relative_path)  
# Converts the relative path into an absolute (full) path based on the current working directory

print(absolute_path)  
# Prints the absolute path to the console
```
