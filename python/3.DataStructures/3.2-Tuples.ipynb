##### Introduction to Tuples in Python
###### tuples are ordered colleecetion of items that are immutable .They are similar to list but the immutability makes it differrent

```python
## creating a tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(lst)
print(type(lst))
```

```python
tpl= tuple()## can add elemennts inside this tuple
print(tpl)
print(type(tpl))
```

```python
numbers=tuple([1,2,3,4,5])## converting list to tuple
print(numbers)
##can also convert tuple to list
```

```python
list((1,2,3,4,5))## converting tuple to list
```

```python
mixed_tuple = (1, "Hello", 3.14, [1,2,3], (4,5), True)
print(mixed_tuple)
```

```python
##assigning values to a tuple
numbers
```

```python
print(numbers[0])## accessing first element of the tuple
print(numbers[1])## accessing second element of the tuple
print(numbers[-1])## accessing last element of the tuple
```

```python
numbers[0:4]## slicing the tuple
```

```python
## Tuples operations
numbers + (6,7,8)## concatenation of tuples
combined_tuple = numbers + mixed_tuple## concatenation of tuples
print(combined_tuple)
```

```python
mixed_tuple*3## repetition of tuples
```

```python
numbers*2## repetition of tuples
```

```python
## immutable nature of tuples
## tuples cannot be modified after creation
lst=[1,2,3]
print(lst)
lst[0]=10## modifying the first element of the list     
print(lst)
tpl=(1,2,3)
print(tpl)
#tpl[0]=10## trying to modify the first element of the tuple (will raise an error)
print(tpl)
```

```python
numbers
```

```python
 ## Tuples methods
 print(numbers.count(1))##how many times the given element is present in the tuple
 print(numbers.index(3))##shows at which index the given element is present
```

```python
## packing and unpacking tuple
## packig
packed_Tuple=1,"hello",3.24
print(packed_Tuple)
type(packed_Tuple)
```

```python
## unpacking a tuple
a,b,c=packed_Tuple
print(a)
print(b)
print(c)

```

```python
## unpacing with star
numbers=(1,2,3,4,5,6)
first,second,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## nested tuples 
lst=[[1,2,3,4],[6,7,8,9],[1,"hello",3.14 ]]
lst[0]
lst[0][3]
lst[0][0:3]
```

```python
lst=[[1,2,3,4],[6,7,8,9],(1,"hello",3.14 )]
lst[2][0:3]
```

```python
nested_tuples=((1,2,3),("a","b","c"),(True,False))
## access the inside a tuple
print(nested_tuples[0])
print(nested_tuples[0][2])
```

```python
## iteratng over nested Tuples
for sub_tuples in nested_tuples:
    for item in sub_tuples:
        print(item,end=" ")
    print()
```
