#### Sets
###### sets are built in python used to Store collection of unique items .They are unordered ,meaning that the element do not follow a specific order,and they do not allow dublicate entries ,and performing mathematical set operataions like union ,intersection ,difference and symmetric difference

```python
### create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(my_empty_set)
```

```python
my_empty_set=set([1,2,3,4,5,6])
print(my_empty_set)
```

```python
## basic set operations 
## adding and remving elements 
my_set.add(7)
print(my_set)
my_set.add(7)##does add or cantainssame element twice
print(my_set)

```

```python
##remove the element from a set 
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(11)
print(my_set)
```

```python
## pop method
removed_element=my_set.pop()
print(removed_element)##what element was first introduced that only will get removed at first
print(my_set)
```

```python
## clear al the elements
my_set.clear()
print(my_set)
```

```python
# set memebership test
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## mathematical operations
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
##Union
union_set=set1.union(set2)
print(union_set)
## instersection
intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)##takes the common elementand puts it in the firstset
print(set1)
```

```python
##
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
##difference
print(set1.difference(set2))
```

```python
## sets method
set1={1,2,3,4,5}
set2={3,4,5}

## is subset
print(set1.issubset(set2))
print(set1.issuperset(set2))
```
