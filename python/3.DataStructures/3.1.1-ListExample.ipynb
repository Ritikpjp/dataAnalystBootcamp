##### Example 1.Manage a TO DO List.

```python
to_do_list=["buy groceries","clean the house","paybills"]
### Adding task 
to_do_list.append("schedulethe meeting")
to_do_list.append("go for a run")

## removing the completed task
to_do_list.remove("clean the house")

## checking if task is in the list or not
if"paybills"in to_do_list:
    print("dont forget to pay the bils")

print("to do list remaining")
for task in to_do_list:
    print(f"-{task}")
```

```python
### organizing student grades
grades=[85,93,56,76,87,45]

## adding new grade
grades.append(90)

## calculating the average grade
average_grade=sum(grades)/len(grades)
print(f"avrage grade:{average_grade:.2f}"
      )
## finding the highest and the lowest grades
highest_grades=max(grades)
lowest_grades=min(grades)
print(f"highest_ grade:{highest_grade}")
print(f"lowest_grades:{lowest_grades}")
```
