```python
##### accessing dictionary element
student={"name":"krish","age":32,"grade":'A'}
print(student)
print(type(student))
```

```python
print(student['grade'])
print(student['age'])
## accesing using get() method 
print(student.get('grade'))
print(student.get('lastname'))
print(student.get('grade',"not in there"))
print(student.get('lastname',"not in there"))

```

```python
### modifying dictionary element
## dictionary are mutable so you can add update or delete elements
print(student)
```

```python
##updating the values in the key
student["age"]=33
print(student)
#adding a new key and value
student["addresss"]="india"
print(student)
```

```python
del student['grade'] ## delete the key and value pair
print(student)
```

```python
## itetrating over dictionaries
## you can use loops to iterate over dictionaries,keys,values,or items
## iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## iterate overvalues
for val in student.values():
    print(val)
```

```python
student.items()
```

```python
## iterate ovver key value pairs
for v in student.items():
    print(v)
```

```python
## iterate ovver key value pairs
for key,val in student.items():
    print(f"{key}:{val}")
```

```python
## nested dictionaries
student={
    "student1":{"name":"krish","age":32},
    "student2":{"name":"spidy","age":31}
}
print(student)
```

```python
## access nested dictionaries element
print(student["student2"]["name"])
print(student["student2"]["age"])
```

```python
student.items()
```

```python
## iterating over nested dictionaris
for student_id,student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## dictionary compprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional dictionary comrehension
evens={x:x**2 for x in range(5) if x%2==0}
print(evens)
```

```python
## use a dictionary to count the freq of element in list
numbers=[1,2,2,3,3,3,4,4,4,4,4]
freq={}
for number in numbers:
    if number in freq:
        freq[number]+=1
    else:
        freq[number]=1
print(freq)

```
