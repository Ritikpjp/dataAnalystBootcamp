```python
## csv
import csv ## this module provides functions for working with CSV files, which are a common format for storing tabular data.
with open('data.csv', mode='w', newline='') as file: ## this will create a new CSV file called 'data.csv' and open it for writing.
    writer = csv.writer(file) ## this will create a CSV writer object that we can use to write data to the file.
    writer.writerow(['Name', 'Age', 'City']) ## this will write a row of data to the CSV file.
    writer.writerow(['Alice', 30, 'New York']) ## this will write another row of data to the CSV file.

with open('data.csv', mode='r') as file: ## this will open the CSV file for reading.
    reader = csv.reader(file) ## this will create a CSV reader object that we can use to read data from the file.
    for row in reader: ## this will iterate over each row in the CSV file.
        print(row) ## this will print each row of data.
```

```python
## datetime
from datetime import datetime,timedelta ## this module provides functions for working with dates and times.
now = datetime.now() ## this will get the current date and time.
print(now) ## this will print the current date and time.    
yesterday = now - timedelta(days=1) ## this will calculate the date and time for yesterday by subtracting one day from the current date and time.
print(yesterday) ## this will print the date and time for yesterday.
```

```python
## time
import time ## this module provides functions for working with time, such as measuring elapsed time and pausing execution.
print(time.time()) ## this will print the current time in seconds since the epoch (January 1, 1970).
time.sleep(2) ## this will pause the execution of the program for 2 seconds.    
print(time.time()) ## this will print the current time again, which should be approximately 2 seconds later than the previous time.
```

```python
##regular expressions
import re ##helps u to match patterns in strings. This module provides functions for working with regular expressions, which are a powerful tool for pattern matching and text manipulation.
pattern = r'\d+' ## this is a regular expression pattern that matches one or more digits.
text = 'The price is 100 dollars 456.' ## this is a string that contains some text and a number.
match = re.search(pattern, text) ## this will search for the pattern in the text and
if match: ## if a match is found, it will return a match object.
    print(match.group()) ## this will print the matched text, which is '100' in this case.
```
