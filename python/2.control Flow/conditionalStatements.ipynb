```python
## if statement
age = 20
if age >= 18:
    print("You are an adult.")
```

```python
## else statement
## the else statement executes a block of code when the condition in the if statement is false.
age = 16
if age >= 18:
    print("You are an adult.")
else:
    print("You are not an adult.")

```

```python
## elif statement
## the elif statement allows you to check multiple conditions after the initial if statement.
## stands for "else if"
age = 65
if age < 18:
    print("You are a minor.")       
elif age < 65:
    print("You are an adult.")      
else:
    print("You are a senior citizen.")
        
```

```python
## nested conditionl statements
## you can nest if statements inside other if statements to create more complex conditions.
## number even ,odd,negative
num=int(input("Enter a number: "))
if num > 0:
    print("The number is positive.")
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
else:
    print("The number is zero or negative.")
    
```

```python
 ## practice example

##you need to determine if the year is a leap year. A leap year is defined as:
## It is divisible by 4 but not by 100, or it is divisible by 400.
year=int(input("Enter a year: "))   
if year % 4 == 0 :
    if year%100==0:
        if year%400==0:
            print(f"{year} is a leap year.")
        else:
            print(f"{year} is not a leap year.")    
    else:
        print(f"{year} is a leap year.")
else:    print(f"{year} is not a leap year.")
    

     
```

```python
##assignment 
# simple calculator program
num1=float(input("Enter first number: "))
num2=float(input("Enter second number: "))              
operation=input("Enter operation (+, -, *, /): ")
if operation == "+":    
    result=num1+num2
    print(f"{num1} + {num2} = {result}")
elif operation == "-":
    result=num1-num2
    print(f"{num1} - {num2} = {result}")
elif operation == "*":
    result=num1*num2
    print(f"{num1} * {num2} = {result}")
elif operation == "/":
    if num2 != 0:
        result=num1/num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Error: Division by zero is not allowed.")
```
