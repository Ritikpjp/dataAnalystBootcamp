```python
## while loop
## the whilwe loop is used to execute a block of code repeatedly as long as a given condition is true
count=0
while count<5:
    print(count)
    count+=1
```

```python
count=0
while count<10:
    if count%2==0:
        print(count)
    count+=1
    # another way to write the above code is
count=0
while count%2==0 or count<10:
    print(count)
    count+=1
```

```python
## loop control  statements
## break statement
## the break statement is used to exit a loop prematurely when a certain condition is met
for i in range(10):
    if i==5:
        break
    print(i)## outsisde the if statement but inside the for loop
print("loop ended")
```

```python
## continue statement
## the continue statements skips the current iteration of the loop and moves to the next iteration
for i in range(10):
    if i<=5:
        continue
        
    print(i)## this will print only odd numbers from 0 to 9
```

```python
## pass
## the pass statement is a null statement that is used as a placeholder when a statement is required syntactically but no action is to be performed
# basically it does nothing and is used when you want to create a loop or a function that you haven't implemented yet but want to avoid syntax errors
for i in range(10):
    if i==3:
        print("three")
        pass
    print(i)## this will print all numbers from 0 to 9 
```

```python
#nested loops
#loops inside loops are called nested loops
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
        print(i,j)
```

```python
## example=calculate thesum of first n natural numbers using a for loop
n=5
total=0
count=1
for i in range(1, n+1):
    total+=i
print(total)
t=0
while count <= n:
    t += count
    count += 1
print(t)
```

```python
## example-Prime numbers between 1 and 100
for num in range(1, 101):
    if num > 1:## prime numbers are greater than 1
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)

```
