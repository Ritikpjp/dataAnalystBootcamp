```python
 ## undersatanding varible types
 ## python is a dynamically typed language,type of variable is detemined at runtime
age=25 # integer
height=5.9 # float
name="John" # string
print("age:",age,"type:",type(age))
```

```python
## type checking and type conversion
type(age) # check the type of variable age
type(height)
```

```python
age=25
print(type(age)) # output: <class 'int'>
# Type conversion
age_str=str(age) # convert integer to string
print(age_str) # output: '25'       
print(type(age_str)) # output: <class 'str'>
```

```python
age='25' # age is now a string
print(type(int(age))) # convert string to integer
```

```python
name="John"
int(name) # this will raise a ValueError because "John" cannot be converted to an integer   
```

```python
height=5.9
type(height) # output: <class 'float'>
```

```python
int(height) # convert float to integer, output: 5   
```

```python
float(height) # convert float to float, output: 5.9
float(int(height)) # convert float to integer and then back to float, output: 5.0
```

```python
## dynamic typing
## python allows the type of a variable to change the program executes
x=10 # x is an integer
print("x:",x,"type:",type(x)) # output: x: 10 type: <class 'int'>
x="Hello" # x is now a string
print("x:",x,"type:",type(x)) # output: x: Hello type: <class 'str'>    
x=3.14 # x is now a float
print("x:",x,"type:",type(x)) # output: x: 3.
```
