##### Indentation
###### indentation in python is used to define hierarchy of the code .unlike many other programming language that use braces {} to delimit a block of code ,python uses indentation to determine the grouping of statements.this means that all the statements whitin the block must be intented at same level.

```python
## Indentation
## In Python, indentation is used to define blocks of code. This is different from many other programming languages that use curly braces or keywords to define blocks. In Python, the level of indentation indicates which statements belong to which blocks of code.
age = 13
if age >= 18:
    print("You are an adult.")

print(age)
```

```python
## line continuation
## use backslash (\) to indicate that a line of code continues on the next line.
#  This is useful for improving readability when you have long lines of code.
total = 1 + 2 + 3 + \
        4 + 5
print(total)
```
