###### -->they determine the type of operation that can be performed onthe data the values that data can take,and the amount of memory nreded to store the data.

```python
## integers example
age=25
type(age) # output: <class 'int'>
```

```python
## floating point data type example
height=5.9
type(height) # output: <class 'float'>
print("Height:",height,"Type:",type(height)) # output: Height: 5.9 Type: <class 'float'>
```

```python
## string data type example
name="John"
print("Name:",name,"Type:",type(name)) # output: Name: John Type: <class 'str'>
```

```python
## boolean datatypes
True
is_student=True
type(is_student) # output: <class 'bool'>
```

```python
a=10
b=10
a==b # output: True, because the values of a and b are equal
type(a==b) # output: <class 'bool'>, because the result of the comparison is a boolean value

```
